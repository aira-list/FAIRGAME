```python
def display_matrix(data):
    weights = data["weights"]
    strategies = data["strategies"]["en"]
    matrix = data["matrix"]

    # Extract strategy names
    strategy_names = list(strategies.values())

    # Determine cell width for proper alignment
    cell_width = 10

    # Header row
    header = " " * cell_width + "|".join(s.center(cell_width) for s in strategy_names)
    print(header)
    print("-" * len(header))

    # Create grid rows
    for i, row_strategy in enumerate(strategy_names):
        row = []
        for j in range(len(strategy_names)):
            combination_key = f"combination{i * len(strategy_names) + j + 1}"
            weight_keys = matrix[combination_key]
            weights_values = [weights[w] for w in weight_keys]
            row.append(f"{weights_values[0]}/{weights_values[1]}".center(cell_width))

        # Print row with the strategy name
        row_content = row_strategy.ljust(cell_width) + "|".join(row)
        print(row_content)
        print("-" * len(header))


# Example data
data = {
    "weights": {"weight1": 3, "weight2": 0, "weight3": 5, "weight4": 1},
    "strategies": {"en": {"strategy1": "Betray", "strategy2": "Cooperate"}},
    "combinations": {
        "combination1": ["strategy1", "strategy1"],
        "combination2": ["strategy1", "strategy2"],
        "combination3": ["strategy2", "strategy1"],
        "combination4": ["strategy2", "strategy2"],
    },
    "matrix": {
        "combination1": ["weight1", "weight1"],
        "combination2": ["weight2", "weight3"],
        "combination3": ["weight3", "weight2"],
        "combination4": ["weight4", "weight4"],
    },
}

# Display the matrix
display_matrix(data)
```
